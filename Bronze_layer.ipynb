```python
files = {"customers": "olist_customers_dataset.csv", 
         "orders": "olist_orders_dataset.csv",
         "order_items": "olist_order_items_dataset.csv",
         "reviews": "olist_order_reviews_dataset.csv",
         "products": "olist_products_dataset.csv",
         "category_translation": "product_category_name_translation.csv"}

for table, filename in files.items():
    path = f"/Volumes/brazilian_ecommerce_dataset_olist/brazilian_ecommerce_dataset_olist_schema/olist_volume/{filename}"
    df = spark.read.format("csv").option("header", True).option("inferSchema", False).option("mode", "PERMISSIVE").option("multiLine", "true").load(path)
    #writes data into Delta tables in Databricks lakehouse
    df.write.format("delta").mode("overwrite").saveAsTable(f"brazilian_ecommerce_dataset_olist.brazilian_ecommerce_dataset_olist_schema.bronze_{table}")  
```
